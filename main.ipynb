```python
# STEP 4: IDENTIFY NUMERICAL AND CATEGORICAL VARIABLES
# ADIM 4: NUMERİK VE KATEGORİK DEĞİŞKENLERİN YAKALANMASI
def grab_col_names(dataframe, cat_th=10, car_th=20):
    # Categorical columns (dtype == object) / Kategorik sütunlar (dtype == object)
    cat_cols = [col for col in dataframe.columns if dataframe[col].dtypes == "O"]
    
    # Numerical columns that should be considered categorical due to low cardinality
    # Düşük benzersizlik nedeniyle kategorik sayılması gereken sayısal sütunlar
    num_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() < cat_th and dataframe[col].dtypes != "O"]
    
    # Categorical columns with high cardinality (too many unique values)
    # Yüksek benzersiz değere sahip kategorik sütunlar
    cat_but_car = [col for col in dataframe.columns if dataframe[col].nunique() > car_th and dataframe[col].dtypes == "O"]

    # Update categorical columns list / Kategorik sütun listesine ekleme
    cat_cols = cat_cols + num_but_cat
    cat_cols = [col for col in cat_cols if col not in cat_but_car]
    
    # Numerical columns (excluding num_but_cat) / Sayısal sütunlar (num_but_cat hariç)
    num_cols = [col for col in dataframe.columns if dataframe[col].dtypes != "O"]
    num_cols = [col for col in num_cols if col not in num_but_cat]
    
    # Print dataset information / Veri seti bilgilerini yazdır
    print(f"Observations: {dataframe.shape[0]}")  # Number of rows / Satır sayısı
    print(f"Variables: {dataframe.shape[1]}")     # Number of columns / Sütun sayısı
    print(f"cat_cols: {len(cat_cols)}")           # Number of categorical columns / Kategorik sütun sayısı
    print(f"num_cols: {len(num_cols)}")           # Number of numerical columns / Sayısal sütun sayısı
    print(f"cat_but_car: {len(cat_but_car)}")     # Number of high cardinality categorical columns / Yüksek benzersizlikte kategorik sütun sayısı
    print(f"num_but_cat: {len(num_but_cat)}")     # Number of numerical columns treated as categorical / Kategorik sayılan sayısal sütun sayısı
    
    return cat_cols, cat_but_car, num_cols

# Call the function to get column lists / Fonksiyonu çağırarak sütun listelerini al
cat_cols, cat_but_car, num_cols = grab_col_names(df)
```

```python
# NUMERICAL VARIABLE ANALYSIS FUNCTION
# SAYISAL DEĞİŞKEN ANALİZİ FONKSİYONU
def num_summary(dataframe, numerical_col, plot=False):
    """
    This function provides a detailed summary of a numerical column in a DataFrame.
    Optionally, it can also plot a histogram of the column values.
    
    Bu fonksiyon, bir DataFrame içindeki sayısal bir sütun için detaylı bir özet sunar.
    İsteğe bağlı olarak, sütun değerlerinin histogramını da çizebilir.
    
    Parameters / Parametreler:
    - dataframe : pandas DataFrame / Analiz yapılacak veri çerçevesi
    - numerical_col : str / Analiz edilecek sayısal sütun adı
    - plot : bool / True ise histogram çizilir, False ise sadece özet gösterilir
    """
    
    # Step 1: Define quantiles to include in the summary
    # Adım 1: Özet istatistiklerde yer alacak çeyrek değerleri tanımla
    quantiles = [0.05, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90, 0.95, 0.99]
    
    # Step 2: Print descriptive statistics including the defined quantiles
    # Adım 2: Tanımlayıcı istatistikleri ve belirlenen çeyrek değerleri yazdır
    # .describe() gives count, mean, std, min, max, and specified quantiles
    # .describe() sütun için sayım, ortalama, standart sapma, min, max ve belirtilen çeyrek değerleri verir
    print(dataframe[numerical_col].describe(quantiles).T)
    
    # Step 3: If plot=True, draw a histogram of the column values
    # Adım 3: Eğer plot=True ise, sütun değerlerinin histogramını çiz
    if plot:
        # Histogram with 50 bins / 50 bölmeli histogram
        dataframe[numerical_col].hist(bins=50)
        
        # Label X-axis with the column name / X eksenini sütun adı ile etiketle
        plt.xlabel(numerical_col)
        
        # Title of the plot is the column name / Grafiğin başlığı sütun adı
        plt.title(numerical_col)
        
        # Display the plot immediately / Grafiği hemen göster
        plt.show(block=True)
    
    # Step 4: Print a separator for clarity in the console output
    # Adım 4: Konsol çıktısında netlik için ayırıcı yazdır
    print("################################")
```

```python
# Loop through all numerical columns and analyze them using num_summary
# Tüm sayısal sütunlar üzerinde döngü oluştur ve num_summary fonksiyonunu kullanarak analiz et
for col in num_cols:
    # Call num_summary for each numerical column / Her sayısal sütun için num_summary fonksiyonunu çağır
    # plot=True to visualize histogram / plot=True ile histogramı görselleştir
    num_summary(df, col, plot=True)
```

```python
# TARGET VARIABLE ANALYSIS FUNCTION
# HEDEF DEĞİŞKEN ANALİZİ FONKSİYONU
def target_summary_with_cat(dataframe, target, categorical_col):
    """
    This function summarizes the target variable (dependent variable) by a categorical feature.
    It calculates and prints the mean of the target variable for each category.
    
    Bu fonksiyon, bir kategorik değişken bazında hedef değişkenin özetini verir.
    Her kategori için hedef değişkenin ortalamasını hesaplar ve yazdırır.
    
    Parameters / Parametreler:
    - dataframe : pandas DataFrame / Analiz yapılacak veri çerçevesi
    - target : str / Hedef değişkenin adı
    - categorical_col : str / Kategorik değişkenin adı
    """
    
    # Group the data by the categorical column and calculate the mean of the target variable
    # Veriyi kategorik sütuna göre gruplandır ve hedef değişkenin ortalamasını hesapla
    summary = pd.DataFrame({
        "TARGET_MEAN": dataframe.groupby(categorical_col)[target].mean()
    })
    
    # Print the summary table / Özet tabloyu yazdır
    # end="\n\n\n" adds extra newlines for readability in console output
    # end="\n\n\n" konsol çıktısında daha okunabilir olması için ekstra satır boşluğu ekler
    print(summary, end="\n\n\n")


# Loop through all categorical columns and analyze the target variable
# Tüm kategorik sütunlar üzerinde döngü oluştur ve hedef değişkeni analiz et
for col in cat_cols:
    # Call the target_summary_with_cat function for each categorical column
    # Her kategorik sütun için target_summary_with_cat fonksiyonunu çağır
    # Here the target variable is 'satisfaction' / Burada hedef değişken 'satisfaction'
    target_summary_with_cat(df, "satisfaction", col)
```

```python
# OUTLIER ANALYSIS FUNCTION
# AYKIRI DEĞER ANALİZİ FONKSİYONU

# This function calculates the lower and upper thresholds for outliers in a numerical variable.
# Bu fonksiyon, bir sayısal değişken için aykırı değerlerin alt ve üst eşiklerini hesaplar.

def outlier_thresholds(dataframe, variable, low_quantile=0.05, up_quantile=0.95):
    """
    Parameters / Parametreler:
    - dataframe : pandas DataFrame / Analiz yapılacak veri çerçevesi
    - variable : str / Aykırı değerleri kontrol edilecek sütun
    - low_quantile : float / Alt çeyrek yüzdelik değeri (default=0.05)
    - up_quantile : float / Üst çeyrek yüzdelik değeri (default=0.95)
    
    Returns / Döndürür:
    - low_limit : float / Alt sınır (alt eşik)
    - up_limit : float / Üst sınır (üst eşik)
    """
    
    # Calculate the lower quantile / Alt çeyrek değerini hesapla
    quantile_one = dataframe[variable].quantile(low_quantile)
    
    # Calculate the upper quantile / Üst çeyrek değerini hesapla
    quantile_three = dataframe[variable].quantile(up_quantile)
    
    # Calculate the interquartile range (IQR) / Çeyrekler arası farkı (IQR) hesapla
    interquantile_range = quantile_three - quantile_one
    
    # Calculate the upper limit for outliers / Aykırı değerler için üst sınır
    up_limit = quantile_three + 1.5 * interquantile_range
    
    # Calculate the lower limit for outliers / Aykırı değerler için alt sınır
    low_limit = quantile_one - 1.5 * interquantile_range
    
    return low_limit, up_limit
```

```python
# OUTLIER CHECK FUNCTION
# AYKIRI DEĞER KONTROL FONKSİYONU

# This function checks whether a numerical column contains outliers
# Bu fonksiyon bir sayısal sütunda aykırı değer olup olmadığını kontrol eder
def check_outlier(dataframe, col_name):
    # Get the lower and upper thresholds for outliers / Aykırı değerler için alt ve üst sınırları al
    low_limit, up_limit = outlier_thresholds(dataframe, col_name)
    
    # Check if any value is below the lower limit or above the upper limit
    # Alt sınırın altında veya üst sınırın üzerinde değer var mı kontrol et
    if dataframe[(dataframe[col_name] > up_limit) | (dataframe[col_name] < low_limit)].any(axis=None):
        return True  # If outliers exist / Aykırı değer varsa True döndür
    else:
        return False  # If no outliers / Aykırı değer yoksa False döndür

# Loop through all numerical columns and check for outliers
# Tüm sayısal sütunlar üzerinde döngü oluştur ve aykırı değerleri kontrol et
# Exclude the target variable "satisfaction" / Hedef değişken "satisfaction" hariç
for col in num_cols:
    if col != "satisfaction":
        # Print the column name and whether it has outliers / Sütun adı ve aykırı değer var mı yazdır
        print(col, check_outlier(df, col))
```

```python
# OUTLIER CAPPING FUNCTION
# AYKIRI DEĞERLERİN BASKILANMASI (CAPPING)

# This function replaces outliers in a numerical column with their corresponding thresholds
# Bu fonksiyon, bir sayısal sütundaki aykırı değerleri belirlenen alt ve üst eşik değerleri ile değiştirir
def replace_with_thresholds(dataframe, variable):
    # Get the lower and upper thresholds for outliers / Aykırı değerler için alt ve üst sınırları al
    low_limit, up_limit = outlier_thresholds(dataframe, variable)
    
    # Replace values below the lower limit with the lower limit / Alt sınırın altındaki değerleri alt sınır ile değiştir
    dataframe.loc[(dataframe[variable] < low_limit), variable] = low_limit
    
    # Replace values above the upper limit with the upper limit / Üst sınırın üzerindeki değerleri üst sınır ile değiştir
    dataframe.loc[(dataframe[variable] > up_limit), variable] = up_limit

# Loop through all numerical columns and apply outlier capping
# Tüm sayısal sütunlar üzerinde döngü oluştur ve aykırı değer baskılamayı uygula
# Exclude the target variable "satisfaction" / Hedef değişken "satisfaction" hariç
for col in num_cols:
    if col != "satisfaction":
        replace_with_thresholds(df, col)

# Re-detect categorical and numerical columns after outlier treatment
# Aykırı değer baskılama sonrası kategorik ve sayısal sütunları yeniden belirle
cat_cols, num_cols, cat_but_car = grab_col_names(df)
```

```python
# LABEL ENCODING FUNCTION
# ETİKET (LABEL) KODLAMA FONKSİYONU

# This function applies label encoding to binary categorical columns
# Bu fonksiyon, ikili kategorik sütunlara label encoding uygular
def label_encoder(dataframe, binary_cols):
    # Fit and transform the binary column using LabelEncoder
    # LabelEncoder kullanarak ikili sütunu sayısal değerlere dönüştür
    dataframe[binary_cols] = LabelEncoder().fit_transform(dataframe[binary_cols])
    return dataframe

# Identify binary categorical columns (2 unique values)
# İkili kategorik sütunları tespit et (2 benzersiz değer)
# NOTE: dtype check "0" seems to be a typo; usually "O" is used for object / dtype kontrolü "0" yazılmış, doğru olan genellikle "O" (object)
binary_cols = [col for col in df.columns if df[col].dtype == "O" and df[col].nunique() == 2]

# Apply label encoding to all identified binary columns
# Tespit edilen tüm ikili sütunlara label encoding uygula
for col in binary_cols:
    df = label_encoder(df, col)

# Display the first 5 rows of the DataFrame after encoding
# Kodlama sonrası DataFrame'in ilk 5 satırını göster
df.head()
```

```python
# ONE-HOT ENCODING FUNCTION
# ONE-HOT ENCODER FONKSİYONU

# This function applies one-hot encoding to categorical columns with more than 2 unique values
# Bu fonksiyon, 2'den fazla benzersiz değere sahip kategorik sütunlara one-hot encoding uygular
def one_hot_encoder(dataframe, categorical_cols, drop_first=True):
    """
    Parameters / Parametreler:
    - dataframe : pandas DataFrame / Analiz yapılacak veri çerçevesi
    - categorical_cols : list / One-hot encoding uygulanacak kategorik sütunlar
    - drop_first : bool / True ise ilk sütun düşürülür (to avoid dummy variable trap / dummy değişken tuzağını önlemek için)
    
    Returns / Döndürür:
    - dataframe : pandas DataFrame with one-hot encoded columns / One-hot encoding uygulanmış DataFrame
    """
    # Use pd.get_dummies to convert categorical columns into dummy variables
    # Kategorik sütunları dummy değişkenlere dönüştür
    dataframe = pd.get_dummies(dataframe, columns=categorical_cols, drop_first=drop_first, dtype=float)
    return dataframe

# Identify categorical columns with more than 2 and at most 11 unique values
# 2'den fazla ve en fazla 11 benzersiz değere sahip kategorik sütunları tespit et
cols = [col for col in df.columns if 11 >= df[col].nunique() > 2]

# Apply one-hot encoding to the identified columns
# Tespit edilen sütunlara one-hot encoding uygula
df = one_hot_encoder(df, cols)

# Display the first 5 rows of the DataFrame after encoding
# Kodlama sonrası DataFrame'in ilk 5 satırını göster
df.head()
```

```python
# SPLIT DATA INTO TRAIN AND TEST SETS
# VERİYİ EĞİTİM VE TEST SETLERİNE BÖLME

# Separate the dataset into training (with target) and test (without target) sets
# Hedef değişkeni olan satırları eğitim seti, hedef değişkeni eksik olan satırları test seti olarak ayır
train_df = df[df["satisfaction"].notnull()]
test_df = df[df["satisfaction"].isnull()]

# Define the target variable (y) and feature matrix (x)
# Hedef değişken (y) ve özellik matrisi (x) olarak ayır
y = train_df["satisfaction"]                 # Target variable / Hedef değişken
x = train_df.drop(["satisfaction"], axis=1)  # Features / Özellikler

# Split the training data into training and validation sets
# Eğitim verisini eğitim ve doğrulama setlerine ayır
# test_size=0.2 → 20% of data for validation / Veri setinin %20'si doğrulama için
# random_state=17 → For reproducibility / Tekrarlanabilirlik için rastgelelik sabitle
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=17)
```

```python
# IMPORT AND DEFINE MACHINE LEARNING CLASSIFIERS
# MAKİNE ÖĞRENMESİ SINIFLANDIRMA MODELLERİNİ TANIMLAMA

# Import classifiers / Sınıflandırıcıları import et
from sklearn.neighbors import KNeighborsClassifier         # K-Nearest Neighbors
from sklearn.linear_model import LogisticRegression        # Logistic Regression
from sklearn.tree import DecisionTreeClassifier            # Decision Tree
from sklearn.ensemble import RandomForestClassifier        # Random Forest
from lightgbm import LGBMClassifier                        # LightGBM
from catboost import CatBoostClassifier                    # CatBoost

# Create a list of model tuples with (name, model instance)
# Model isimleri ve instance’larını içeren bir liste oluştur
# This allows looping through models for training and evaluation
# Bu liste, modeller üzerinde döngü kurarak eğitim ve değerlendirme yapmayı kolaylaştırır
models = [
    ("KNN", KNeighborsClassifier()),                              # K-Nearest Neighbors classifier / K-en yakın komşu sınıflandırıcısı
    ("Logistic Regression", LogisticRegression(max_iter=1000)),   # Logistic Regression / Lojistik regresyon (maks iterasyon 1000)
    ("CART", DecisionTreeClassifier()),                            # Decision Tree classifier / Karar ağacı sınıflandırıcısı
    ("Random Forest", RandomForestClassifier()),                  # Random Forest / Rastgele orman
    ("LightGBM", LGBMClassifier(verbose=-1)),                     # LightGBM classifier (verbose=-1 siler logları) / LightGBM sınıflandırıcı
    ("CatBoost", CatBoostClassifier(verbose=False))               # CatBoost classifier (verbose=False siler logları) / CatBoost sınıflandırıcı
]
```

```python
# LABEL ENCODING FOR TRUE CATEGORICAL VARIABLES
# GERÇEK KATEGORİK DEĞİŞKENLER İÇİN LABEL ENCODING

from sklearn.preprocessing import LabelEncoder

# List of categorical columns that need encoding
# Encode edilmesi gereken gerçek kategorik sütunlar
true_categorical_cols = ['Gender', 'Customer Type', 'Type of Travel']

# Create copies of training and test sets to preserve original data
# Orijinal veriyi korumak için kopyalar oluştur
x_train_encoded = x_train.copy()
x_test_encoded = x_test.copy()

# Dictionary to store LabelEncoders for each column
# Her sütun için LabelEncoder’ları saklayacak sözlük
label_encoders = {}

# Apply Label Encoding to each categorical column
# Her kategorik sütuna Label Encoding uygula
for col in true_categorical_cols:
    le = LabelEncoder()
    
    # Fit encoder on training data and transform both training and test sets
    # Encoder’ı eğitim verisine fit et ve eğitim & test verisini dönüştür
    x_train_encoded[col] = le.fit_transform(x_train[col])
    x_test_encoded[col] = le.transform(x_test[col])
    
    # Store the fitted encoder in dictionary for future use
    # Fit edilmiş encoder’ı sözlükte sakla
    label_encoders[col] = le

# Print mapping of original categories to encoded numbers
# Orijinal kategorilerin sayısal karşılıklarını yazdır
print("Kategorik değişkenler encode edildi:")
for col in true_categorical_cols:
    le = label_encoders[col]
    print(f"{col}: {dict(zip(le.classes_, range(len(le.classes_))))}")
```

```python
# FEATURE IMPORTANCE PLOTTING FUNCTION
# ÖZELLİK ÖNEMİ GÖRSELLEŞTİRME FONKSİYONU

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_feature_importance(importance, names, model_type):
    """
    This function plots the feature importance of a model.
    It also prints the top 10 most important features.
    
    Bu fonksiyon, bir modelin feature importance (özellik önemini) görselleştirir.
    Ayrıca en önemli 10 özelliği yazdırır.
    
    Parameters / Parametreler:
    - importance : list or array / Model tarafından belirlenen feature importance değerleri
    - names : list / Özellik isimleri
    - model_type : str / Model adı (ör. Random Forest, LightGBM)
    """
    
    # Convert importance and feature names into numpy arrays / Değerleri numpy array'e dönüştür
    feature_importance = np.array(importance)
    feature_names = np.array(names)
    
    # Create a DataFrame for easy sorting and plotting / Kolay sıralama ve görselleştirme için DataFrame oluştur
    data = {"feature_names": feature_names, "feature_importance": feature_importance}
    fi_df = pd.DataFrame(data)
    
    # Sort features by importance in descending order / Özellikleri önemine göre azalan sırada sırala
    fi_df.sort_values(by=["feature_importance"], ascending=False, inplace=True)
    
    # Plot the feature importance / Feature importance grafiğini çiz
    plt.figure(figsize=(12, 10))
    sns.barplot(x=fi_df["feature_importance"], y=fi_df["feature_names"])
    
    plt.title(model_type + " FEATURE IMPORTANCE")
    plt.xlabel("FEATURE IMPORTANCE")
    plt.ylabel("FEATURE NAMES")
    plt.tight_layout()
    plt.show()
    
    # Print top 10 important features / En önemli 10 feature'ı yazdır
    print(f"\n{model_type} - EN ÖNEMLİ 10 FEATURE:")
    print(fi_df.head(10)[['feature_names', 'feature_importance']].to_string(index=False))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# 1️⃣ Selected models for feature importance analysis
# ÖNEMLİ: Feature importance analizi için seçilen modeller
selected_models = [
    ("CART", DecisionTreeClassifier(random_state=42)),              # Tekil karar ağacı
    ("RF", RandomForestClassifier(random_state=42, n_estimators=100)),  # Random Forest (100 ağaç)
    ("LightGBM", LGBMClassifier(random_state=42, verbose=-1)),      # LightGBM, loglar kapalı
    ("CatBoost", CatBoostClassifier(random_state=42, verbose=False)),  # CatBoost, loglar kapalı
]

print("FEATURE IMPORTANCE ANALİZİ BAŞLIYOR...")
print("=" * 50)

# 2️⃣ Loop through each selected model
# Her bir seçilmiş model üzerinde döngü
for name, model in selected_models:
    print(f"\n{name} modeli eğitiliyor...")
    
    try:
        # 3️⃣ Fit the model on training data
        # Modeli eğitim verisi ile eğit
        model.fit(x_train_encoded, y_train)
        print(f"{name} modeli başarıyla eğitildi!")
        
        # 4️⃣ Get feature importance
        # Modelin feature importance değerlerini al
        if hasattr(model, 'feature_importances_'):
            # RandomForest, DecisionTree, LightGBM, CatBoost gibi modellerde bulunur
            importance = model.feature_importances_
        elif hasattr(model, 'coef_'):
            # LogisticRegression veya lineer modeller için katsayıların mutlak değeri
            importance = np.abs(model.coef_[0])
        else:
            print(f"{name} modelinde feature_importances_ bulunamadı!")
            continue
        
        # 5️⃣ Plot feature importance using the defined function
        # Daha önce tanımlanan plot_feature_importance fonksiyonunu kullanarak grafiği çiz
        plot_feature_importance(importance, x_train_encoded.columns.tolist(), model_type=name)
        
        # 6️⃣ Show model performance on training and validation sets
        # Eğitim ve doğrulama skorlarını yazdır
        train_score = model.score(x_train_encoded, y_train)
        test_score = model.score(x_test_encoded, y_test)
        print(f"{name} Train Score: {train_score:.4f}")
        print(f"{name} Test Score: {test_score:.4f}")
        
    except Exception as e:
        # 7️⃣ Error handling
        # Eğer model eğitiminde hata olursa yazdır
        print(f"{name} modelinde hata: {str(e)}")
    
    print("-" * 50)

print("\nFEATURE IMPORTANCE ANALİZİ TAMAMLANDI!")
```

```python
# ================================
# FEATURE IMPORTANCE ANALYSIS PIPELINE
# ÖZELLİK ÖNEMİ ANALİZİ PIPELINE
# ================================

# 1️⃣ Gerekli kütüphaneler import edildi
# Numpy, Pandas → Veri manipülasyonu ve sayısal işlemler / Data manipulation & numerical operations
# Matplotlib, Seaborn → Görselleştirme ve grafikler / Data visualization
# Sklearn, LightGBM, CatBoost → Modeller ve preprocessing / Models and preprocessing
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder          # Kategorik değişkenleri sayısala çevirir / Converts categorical columns to numerical
from sklearn.tree import DecisionTreeClassifier         # Tek karar ağacı sınıflandırıcı / Single decision tree classifier
from sklearn.ensemble import RandomForestClassifier     # Random Forest sınıflandırıcı / Ensemble of decision trees
from lightgbm import LGBMClassifier                     # LightGBM sınıflandırıcı / Fast gradient boosting classifier
from catboost import CatBoostClassifier                 # CatBoost sınıflandırıcı / Gradient boosting for categorical features

# 2️⃣ Kategorik değişkenleri encode et
# Gerçek kategorik sütunlar / True categorical columns
true_categorical_cols = ['Gender', 'Customer Type', 'Type of Travel']

# Eğitim ve test verilerinin kopyaları oluşturuluyor (orijinal veri korunur)
x_train_encoded = x_train.copy()
x_test_encoded = x_test.copy()

# Her kategorik sütun için LabelEncoder uygula ve encoder'ı sakla
# LabelEncoder, kategorik değerleri 0,1,2 gibi sayısal değerlere dönüştürür
label_encoders = {}
for col in true_categorical_cols:
    le = LabelEncoder()
    # Eğitim verisini fit et ve dönüştür
    x_train_encoded[col] = le.fit_transform(x_train[col])
    # Test verisini aynı mapping ile dönüştür
    x_test_encoded[col] = le.transform(x_test[col])
    # Encoder'ı sakla, ileride decode veya yeni veri için kullanılabilir
    label_encoders[col] = le

# 3️⃣ Feature importance görselleştirme fonksiyonu
def plot_feature_importance(importance, names, model_type):
    """
    Parameters / Parametreler:
    - importance : Modelin feature importance değerleri
    - names : Özellik isimleri / Feature names
    - model_type : Model adı (ör. Random Forest)
    
    Fonksiyon:
    - Feature importance değerlerini sıralar
    - Barplot ile görselleştirir
    - En önemli 10 feature’ı yazdırır
    """
    
    # Numpy array'e çevir / Convert to numpy arrays
    feature_importance = np.array(importance)
    feature_names = np.array(names)
    
    # DataFrame oluştur, sıralama ve görselleştirme için / Create DataFrame for sorting & plotting
    data = {"feature_names": feature_names, "feature_importance": feature_importance}
    fi_df = pd.DataFrame(data)
    
    # Feature'ları önem sırasına göre sırala / Sort features by importance
    fi_df.sort_values(by=["feature_importance"], ascending=False, inplace=True)
    
    # Barplot oluştur / Create barplot
    plt.figure(figsize=(12, 10))  # Grafiğin boyutu / Figure size
    sns.barplot(x=fi_df["feature_importance"], y=fi_df["feature_names"])
    plt.title(model_type + " FEATURE IMPORTANCE")  # Başlık / Title
    plt.xlabel("FEATURE IMPORTANCE")               # X ekseni etiketi / X-axis label
    plt.ylabel("FEATURE NAMES")                    # Y ekseni etiketi / Y-axis label
    plt.tight_layout()                             # Layout düzeltmesi / Adjust layout
    plt.show()                                     # Grafiği göster / Show plot
    
    # En önemli 10 feature'ı yazdır / Print top 10 features
    print(f"\n{model_type} - EN ÖNEMLİ 10 FEATURE:")
    print(fi_df.head(10)[['feature_names', 'feature_importance']].to_string(index=False))

# 4️⃣ Modelleri seç / Define models
selected_models = [
    ("CART", DecisionTreeClassifier(random_state=42)),                  # Tek karar ağacı / Single decision tree
    ("RF", RandomForestClassifier(random_state=42, n_estimators=100)),  # Random Forest / Ensemble of trees
    ("LightGBM", LGBMClassifier(random_state=42, verbose=-1)),          # LightGBM / Fast gradient boosting
    ("CatBoost", CatBoostClassifier(random_state=42, verbose=False)),   # CatBoost / Gradient boosting with categorical handling
]

# 5️⃣ Feature importance analizi döngüsü / Loop through models
print("FEATURE IMPORTANCE ANALİZİ BAŞLIYOR...")
print("=" * 50)

for name, model in selected_models:
    print(f"\n{name} modeli eğitiliyor... / Training {name} model...")
    
    try:
        # Modeli eğitim verisi ile fit et / Fit the model on training data
        model.fit(x_train_encoded, y_train)
        print(f"{name} modeli başarıyla eğitildi! / {name} model trained successfully!")
        
        # Feature importance değerlerini al / Get feature importance values
        importance = model.feature_importances_
        
        # Feature importance grafiğini çiz / Plot feature importance
        plot_feature_importance(importance, x_train_encoded.columns.tolist(), name)
        
        # Eğitim ve test skorlarını hesapla ve yazdır / Calculate & print train/test scores
        train_score = model.score(x_train_encoded, y_train)
        test_score = model.score(x_test_encoded, y_test)
        print(f"{name} Train Score: {train_score:.4f} / Test Score: {test_score:.4f}")
        
    except Exception as e:
        # Hata yakalama / Catch any errors
        print(f"{name} modelinde hata: {str(e)} / Error in {name} model: {str(e)}")
    
    print("-" * 50)

print("\nANALİZ TAMAMLANDI! 🎉 / ANALYSIS COMPLETED! 🎉")
```
